### 1. Line Chart: Average Daily Social Media Hours by Age

```python
avg_stress = (df.groupby("age")["stress_level"].mean())

plt.plot(avg_stress.index,
    avg_stress.values,
    marker="o",
        color = "red")

plt.title("Average Stress Level by Age")
plt.xlabel("Age")
plt.ylabel("Stress Level")

plt.grid(True)

plt.show()
```

**Inisights:** Average stress levels show that on avg, people of the age 14 tend to stress more than others. Then it drops down dramatically during mid-adolesence 

### 2. Bar Chart
- #### Primary Platform usage

```python
plt.bar(df["platform_usage"].value_counts().index,
        df["platform_usage"].value_counts().values,
       color = "purple")
```

**Insights**: The most used platform among the teens seems to be Instagram followed by Tiktok

- #### Depression Label Distribution

```python
depression = df["depression_label"].value_counts()

plt.figure(figsize=(6,5))

plt.bar(
    ["No Depression", "Depression"],
    depression.values
)

plt.title("Depression Label Distribution")
plt.xlabel("Category")
plt.ylabel("No. of Teenagers")

plt.show()
```

**Insights:** Most teens fall into one class. Most of them were diagnosed with no depression

### 3. Histogram of Daily Social Media Hours Spent

```python
plt.hist(df["daily_social_media_hours"],
        bins = 30, color = "lightgreen")
plt.title("Daily Social Media Hours Spent")
plt.xlabel("Social Media Hours per Day")
plt.ylabel("Frequency")
```

**Insights:** Daily social media hrs are pretty much uniformly distrbuted. There is no distinct peak 

### 4. Boxplot for Anxiety Levels

```python
plt.figure(figsize=(6,5))

plt.boxplot(df["anxiety_level"])

plt.title("Boxplot of Anxiety Level")
plt.ylabel("Anxiety Level")

plt.show()
```

**Insights:** There seem to be no outliers

### 5. Scatterplot : Daily Social Media Hours vs Academic Performance

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["daily_social_media_hours"],
    df["academic_performance"],
    alpha=0.8
)

plt.title("Daily Social Media Hours vs Academic Performance")
plt.xlabel("Daily Social Media Hours")
plt.ylabel("Academic Performance")

plt.show()
```

**Insights**: The above scatter pklot shows no strong relation betweeen the daily usafe and academic performance, that means some other factors havea greater influence

### 6. Heatmap

```python
corr_columns = ['age', 'daily_social_media_hours', 'sleep_hours', 'stress_level', 'anxiety_level', 'depression_label']
corr_matrix = df[corr_columns].corr().values

plt.figure(figsize=(6, 5))
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Correlation Matrix")

plt.xticks(range(len(corr_columns)), corr_columns, rotation=90)
plt.yticks(range(len(corr_columns)), corr_columns)

plt.colorbar()
plt.show()

```

**Insights**: There is a mild neg correlation of about-0.19 to -0.20 that show decreased sleep hours track with elevated depression indicators.
Surprisingly the other ones seem to exist independently, Maybe the dataset is the issue T-T

### Questions
1. When should you use a bar chart vs. a histogram?\
Use a bar chart to compare categories and a histogram to see how numerical data is distributed
2. What does a boxplot show that a histogram doesn’t?\
A boxplot shows the median, spread, and any outliers in the data
3. How do you add labels, title, and legend to a Matplotlib chart?\
we use plt.xlabel(), plt.ylabel(), plt.title(), and plt.legend()
4. What’s a heatmap typically used for in EDA?\
A heatmap is used to see how different numerical variables are related to each other.
5. Why is a pie chart usually a bad choice for data with more than 4–5 categories?\
Because it gets crowded and makes it hard to compare the different categories.
