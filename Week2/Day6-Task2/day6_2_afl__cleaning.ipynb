# 2. Cleaning
### `info_df`
#### Decisions
1. **Drop exact duplicate rows**
2. **`weight == 0` → `NaN`** : zero kilograms is not a real value for an AFL player; treating it as missing avoids polluting any weight-based analysis while preserving the rest of the record.
4. **Parse date columns to `datetime`**:  enables correct date arithmetic and consistent downstream formatting.
5. **Leave `profile_pic`, `player_common_names`, `player_teams` missing values as `NaN`** : these are optional/metadata fields; imputing them would fabricate data (e.g. inventing a nickname or team history).
6. **No changes to height/weight extremes or date logic** : verified as genuine values, not errors.


```python
info = info_df.copy()

# 1. Remove exact duplicate rows
dup_mask = info.duplicated()
n_dup = dup_mask.sum()
info = info[~dup_mask].copy()
print(f"Removed {n_dup} exact duplicate rows.")

# 2. weight == 0 -> NaN (impossible value)
n_bad_weight = (info['weight'] == 0).sum()
info.loc[info['weight'] == 0, 'weight'] = np.nan
print(f"Set {n_bad_weight} weight==0 values to NaN.")

# 3. Parse dates
for c in ['born_date', 'debut_date', 'last_date']:
    info[c] = pd.to_datetime(info[c], errors='coerce')
print("Converted born_date, debut_date, last_date to datetime dtype.")

# 4. Ensure id is a clean string key for merging later
info['id'] = info['id'].astype(str)

info_rows_after = len(info)
info_rows_before = len(info_df)
print(f"Rows before {info_rows_before}, and Rows after {info_rows_after}")
```

```python
info.isnull().sum()
```

### `seasonal_stats`

**Decisions:**
1. **Normalize `player_id`** — strip the `ID_` prefix from the 10 affected rows so every id is a plain numeric string matching `players_info.id`. Without this, those 10 rows would silently fail to merge.
2. **Drop exact duplicate rows** :identical player/year/team/stat combinations are double-entries, not separate seasons.
3. **Fix negative `games_played`**: a player cannot play a negative number of games. Taking the absolute value corrects what appears to be a sign-entry error (all other stats in these rows are positive and consistent with the corrected count).
4. **Standardize `team` names** :collapse case/whitespace variants (`' GEELONG CATS '`, `'geelong cats'`, `'Geelong Cats'`) into one canonical name matching the team naming convention used in `players_info`.
5. **Leave negative `total_fantasy_points` unchanged** :legitimate under AFL fantasy scoring (clangers, turnovers, and free kicks against subtract points), confirmed by the fact it only occurs in low-game-time rows.
6. **Leave missing advanced-stat values as `NaN`** : these reflect genuine historical non-tracking (e.g. hit-outs and Brownlow votes were not consistently recorded league-wide in earlier seasons). Imputing them (e.g. with 0 or mean) would fabricate performance data and bias any era-comparison analysis.


```python
TEAM_MAP = {
    'adelaide crows': 'Adelaide Crows', 'brisbane bears': 'Brisbane Bears',
    'brisbane lions': 'Brisbane Lions', 'carlton blues': 'Carlton Blues',
    'collingwood magpies': 'Collingwood Magpies', 'essendon bombers': 'Essendon Bombers',
    'fitzroy lions': 'Fitzroy Lions', 'fremantle dockers': 'Fremantle Dockers',
    'geelong cats': 'Geelong Cats', 'gold coast suns': 'Gold Coast Suns',
    'greater western sydney giants': 'Greater Western Sydney Giants',
    'hawthorn hawks': 'Hawthorn Hawks', 'melbourne demons': 'Melbourne Demons',
    'north melbourne kangaroos': 'North Melbourne Kangaroos',
    'port adelaide power': 'Port Adelaide Power', 'richmond tigers': 'Richmond Tigers',
    'st kilda saints': 'St Kilda Saints', 'sydney swans': 'Sydney Swans',
    'west coast eagles': 'West Coast Eagles', 'western bulldogs': 'Western Bulldogs',
}

stats = stats_df.copy()

# 1. Normalize player_id (strip 'ID_' prefix)
n_prefixed = stats['player_id'].astype(str).str.contains('ID_').sum()
stats['player_id'] = stats['player_id'].astype(str).str.replace(r'\D', '', regex=True)
print(f" Normalized player_id: stripped 'ID_' prefix from {n_prefixed} values.")

# 2. Remove exact duplicate rows
dup_mask_s = stats.duplicated()
n_dup_s = dup_mask_s.sum()
stats = stats[~dup_mask_s].copy()
print(f" Removed {n_dup_s} exact duplicate rows.")

# 3. Fix negative games_played (sign error)
n_neg_games = (stats['games_played'] < 0).sum()
stats.loc[stats['games_played'] < 0, 'games_played'] = stats.loc[stats['games_played'] < 0, 'games_played'].abs()
print(f" Fixed {n_neg_games} rows with negative games_played (took absolute value).")

# 4. Standardize team names
before_unique = stats['team'].nunique()
stats['team'] = stats['team'].str.strip().str.lower().map(TEAM_MAP).fillna(stats['team'].str.strip())
after_unique = stats['team'].nunique()
print(f"Standardized team names: {before_unique} raw variants -> {after_unique} canonical teams.")

# 5. total_fantasy_points negatives -> left unchanged (legitimate)
n_neg_fp = (stats['total_fantasy_points'] < 0).sum()
print(f"total_fantasy_points negative in {n_neg_fp} rows - left unchanged (valid fantasy scoring outcome).")

stats_rows_after = len(stats)
stats_rows_before = len(stats_df)
print(f"Rows before {stats_rows_before}, and Rows after {stats_rows_after}")
```

```python
missing_summary = stats.isnull().sum()
missing_summary = missing_summary[missing_summary > 0]
print(f"[stats] {len(missing_summary)} columns retain missing values:- left as NaN")
missing_summary.sort_values(ascending=False)
```

```python
count_cols_with_na = [c for c in stats.columns 
                       if stats[c].isnull().sum() > 0 and not c.startswith('avg_')]

stats[count_cols_with_na] = stats[count_cols_with_na].fillna(0)

# Recompute the matching avg_ column from the (now-filled) count and games_played,rather than filling avg_ columns independently
for col in count_cols_with_na:
    avg_col = f'avg_{col}'
    if avg_col in stats.columns:
        stats[avg_col] = (stats[col] / stats['games_played']).round(1)

remaining_avg_na = [c for c in stats.columns if c.startswith('avg_') and stats[c].isnull().sum() > 0]
stats[remaining_avg_na] = stats[remaining_avg_na].fillna(0)

print(f"Filled {len(count_cols_with_na)} count columns and recomputed their averages.")
print(f"Filled {len(remaining_avg_na)} remaining avg-only columns with 0.")
print("Remaining NaNs in stats:", stats.isnull().sum().sum())
```

## 3. Merge & Unmatched-ID Analysis

Players are merged on `seasonal_stats.player_id` = `players_info.id`. We use an
**inner join** for the analysis-ready `merged_players.csv` (every row has both
season performance and player biographical data), and separately report every
`player_id` that failed to match so no data loss is hidden.

```python
info_ids = set(info['id'])
stats_ids = set(stats['player_id'])

unmatched_in_stats = stats_ids - info_ids   # season-stat ids with no player record
unmatched_in_info = info_ids - stats_ids    # players with no season stats at all

print(f"Unique player ids - stats: {len(stats_ids)}, info: {len(info_ids)}")
print(f"player_id in seasonal_stats with NO match in players_info: {len(unmatched_in_stats)}")
print(f"id in players_info with NO season stats recorded: {len(unmatched_in_info)}")
```

```python
merged = stats.merge(info, left_on='player_id', right_on='id', how='inner', suffixes=('', '_info'))
merged = merged.drop(columns=['id'])  # redundant with player_id

n_merged = len(merged)
print(f"row count: {n_merged} "
    f"({stats_rows_after - n_merged} stats rows excluded :unmatched player_id).")

merged.head()
```

```python
unmatched_ids_df = pd.DataFrame(sorted(unmatched_in_stats), columns=['unmatched_player_id'])
print(f"Sample of unmatched player_id values ({len(unmatched_ids_df)} total):")
unmatched_ids_df.head(20)
```
